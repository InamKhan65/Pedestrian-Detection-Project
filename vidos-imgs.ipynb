```python
import cv2
import os

# ---- SETTINGS ----
video_path = 'D:/saadu/fyp/Projects/inam/PVideos/2.mp4'
output_dir = 'D:/saadu/fyp/Projects/inam/PVideos/Imgs2'
frame_skip = 20  # Save every 20th frame

# ---- SETUP ----
os.makedirs(output_dir, exist_ok=True)
cap = cv2.VideoCapture(video_path)

frame_count = 0
saved_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    if frame_count % frame_skip == 0:
        output_path = os.path.join(output_dir, f'frame_{saved_count:05d}.jpg')
        cv2.imwrite(output_path, frame)
        saved_count += 1

    frame_count += 1

cap.release()
print(f"Extracted {saved_count} frames into '{output_dir}'")
```
